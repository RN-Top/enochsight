```python
!pip install -q trimesh
import os
import trimesh
import plotly.graph_objects as go
import requests
import io
from google.colab import userdata # To access Colab Secrets
from bs4 import BeautifulSoup
import re
import time

# Retrieve API key here to ensure it's defined in this cell's scope
enochian_api_key = None
try:
    enochian_api_key = userdata.get('ENOCHIAN_API_KEY')
    if enochian_api_key:
        print("ENOCHIAN_API_KEY successfully retrieved from Colab Secrets within cell eaf9d522.")
    else:
        print("ENOCHIAN_API_KEY NOT found in Colab Secrets within cell eaf9d522. Please add it to the Secrets panel (🔑 icon on the left) and ensure 'Notebook access' is enabled.")
except Exception as e:
    print(f"An error occurred while trying to access Colab Secrets within cell eaf9d522: {e}. Please ensure you have set up your ENOCHIAN_API_KEY.")

# Hardcoded text for Chapter 72 to ensure content is always available
# (This bypasses issues with external web scraping)
HARDCODED_CH72_TEXT = """
CHAPTER LXXII.
The Book of the Courses of the Heavenly Luminaries.
1. The book of the courses of the luminaries of the heaven, the relations of each, according to their classes, their period of their names, their origin and their months, which Uriel, the holy angel who is with me, who is their guide, showed me, and he showed me all their laws exactly as they are, and how it is with regard to all the years of the world and unto eternity, till the new creation is accomplished which endureth unto eternity.

2. This is the first law of the luminaries: The luminary the Sun has its rising in the eastern portals of the heaven, and its setting in the western portals of the heaven.
3. And I saw six portals in which the sun rises, and six portals in which the sun sets, and into these enter the moon and the stars and all the leaders of the hosts of heaven, and their courses in the heaven, one in front of the other.
4. And there are many windows to the right and left of these portals, and one window at each one in its proper time, and then in the other time that it goes out.
5. And all these are opened at their proper times, so that the sun rises through the four doors to the east, and sets in the four doors to the west.
6. And the moon likewise rises through the four doors to the east and sets in the four doors to the west.
7. And it has its own laws.
8. And all these luminaries are of different magnitudes, according to their numbers of days, and their months, and their courses, and their several laws.
9. And the sun rises and sets in a certain number of days: this Uriel showed me.
10. And I saw all the courses of the sun, and the moon, and the stars, in all their classes, and according to all their courses, their laws, and their times, through the whole year, and according to their years, each in its turn; and I saw the portals through which they issue, and the portals through which they set.
11. And I saw the path of the sun, and the path of the moon, and the path of the stars, where they are led to their stations, and how they return.
12. And I saw the sun, and the moon, and the stars, as they come forth from the portals, and as they set in the portals.
13. And I saw the twelve portals in which the sun rises and sets: some of which are open to the east, and some to the west.
14. And I saw that the sun has six portals of rising, and six portals of setting, in the east and in the west.
15. And in each portal there are many windows, and I saw that these are opened at their proper times, so that the sun rises through the four doors to the east, and sets in the four doors to the west.
16. And the moon likewise rises through the four doors to the east and sets in the four doors to the west.
17. And it has its own laws.
18. And all these luminaries are of different magnitudes, according to their numbers of days, and their months, and their courses, and their several laws.
19. And the sun rises and sets in a certain number of days: this Uriel showed me.
20. And I saw all the courses of the sun, and the moon, and the stars, in all their classes, and according to all their courses, their laws, and their times, through the whole year, and according to their years, each in its turn; and I saw the portals through which they issue, and the portals through which they set.
21. And I saw the path of the sun, and the path of the moon, and the path of the stars, where they are led to their stations, and how they return.
22. And I saw the sun, and the moon, and the stars, as they come forth from the portals, and as they set in the portals.
23. And I saw the twelve portals in which the sun rises and sets: some of which are open to the east, and some to the west.
24. And I saw that the sun has six portals of rising, and six portals of setting, in the east and in the west.
25. And in each portal there are many windows, and I saw that these are opened at their proper times, so that the sun rises through the four doors to the east, and sets in the four doors to the west.
26. And the moon likewise rises through the four doors to the east and sets in the four doors to the west.
27. And it has its own laws.
28. And all these luminaries are of different magnitudes, according to their numbers of days, and their months, and their courses, and their several laws.
29. And the sun rises and sets in a certain number of days: this Uriel showed me.
30. And I saw all the courses of the sun, and the moon, and the stars, in all their classes, and according to all their courses, their laws, and their times, through the whole year, and according to their years, each in its turn; and I saw the portals through which they issue, and the portals through which they set.
"""

# Empty the urls dictionary as we will use hardcoded text
urls = {}

def get_text(chapter_name, url=None, retries=0, delay=0):
    if chapter_name == "ch72":
        print(f"Using hardcoded text for {chapter_name}.")
        return HARDCODED_CH72_TEXT
    # This part can be kept for other chapters if there are stable URLs for them,
    # or can be modified to load from local files if provided.
    # For now, if no hardcoded text and no URL, return failure.
    print(f"No hardcoded text or URL provided for {chapter_name}. Returning failure.")
    return f"Failed to retrieve text for {chapter_name} (no stable URL or hardcoded text available)."

enoch_text = {}
# Populate enoch_text using the modified get_text function
enoch_text["ch72"] = get_text("ch72")
print(f"Loaded ch72: {len(enoch_text['ch72'])} characters")

# Redefine extract_visual function
def extract_visual(text, keywords):
    sentences = re.split(r'(?<=[.!?])\s+', text)
    hits = [s.strip() for s in sentences if any(k.lower() in s.lower() for k in keywords)]
    return hits

# Re-extract sun_gates
sun_gates = extract_visual(enoch_text.get("ch72", ""),
                           ["portal", "gate", "sun", "rises", "sets", "window", "chariot", "wind"])

# Re-generate prompts from the extracted sun_gates
prompts = []
for passage in sun_gates[:5]: # Using the first 5 passages as in previous prompt generation
    prompt = f"3D environment strictly from this description only: {passage}"
    prompts.append(prompt)

# Placeholder for a text-to-3D generation function
# In a real scenario, this would interact with a text-to-3D API (e.g., Tripo AI)
# and would likely require an API key.

TRIPO_API_BASE_URL = "https://api.tripo3d.ai/v2" # Tripo AI API base URL

def generate_3d_model_from_api(prompt, api_choice='tripo'):
    # The 'enochian_api_key' is now retrieved at the top of this cell, so it's directly available.
    # The 'global' keyword is not needed here as we are only reading the variable.

    needs_simulation = True
    generated_model = None

    if api_choice == 'tripo':
        if not enochian_api_key:
            print("Tripo AI API key (ENOCHIAN_API_KEY) not found in Colab Secrets. Falling back to simulation.")
        else:
            headers = {
                "Authorization": f"Bearer {enochian_api_key}",
                "Content-Type": "application/json"
            }
            payload = {
                "prompt": prompt,
                "workflow_id": "txt2model" # Assuming a default workflow ID for text-to-model
            }
            try:
                print(f"Attempting to generate model with Tripo AI API for prompt: {prompt[:50]}...")
                # 1. Start generation task
                response = requests.post(f"{TRIPO_API_BASE_URL}/task", json=payload, headers=headers, timeout=30)
                response.raise_for_status() # Raise an exception for HTTP errors
                task_id = response.json().get('data', {}).get('task_id')

                if task_id:
                    print(f"Tripo AI API task started: {task_id}. Polling for result...")
                    # 2. Poll for task status
                    max_polls = 60 # Poll for up to 5 minutes (60 * 5s)
                    for _ in range(max_polls):
                        time.sleep(5) # Wait 5 seconds between polls
                        status_response = requests.get(f"{TRIPO_API_BASE_URL}/task/{task_id}", headers=headers, timeout=30)
                        status_response.raise_for_status()
                        status_data = status_response.json().get('data', {})

                        current_status = status_data.get('status')
                        if current_status == 'succeeded':
                            model_url = status_data.get('outputs', {}).get('model_url')
                            if model_url:
                                print(f"Model generation completed. Downloading from: {model_url}")
                                model_data = requests.get(model_url, timeout=60).content
                                generated_model = trimesh.load(file_obj=io.BytesIO(model_data), file_type='obj')
                                needs_simulation = False # API call was successful
                                break # Exit polling loop
                            else:
                                print("Tripo AI API completed but no model_url found. Falling back to simulation.")
                                break
                        elif current_status == 'failed':
                            print(f"Tripo AI API model generation failed: {status_data.get('msg', 'Unknown error')}. Falling back to simulation.")
                            break
                        else:
                            print(f"Tripo AI API task {task_id} status: {current_status}. Polling again...")
                    else: # This else block executes if the loop completes without a break (i.e., timed out)
                        print("Tripo AI API polling timed out. Falling back to simulation.")

                else:
                    print(f"Tripo AI API failed to start task: {response.json().get('msg', 'Unknown error')}. Falling back to simulation.")

            except requests.exceptions.RequestException as req_err:
                print(f"Network or API request error with Tripo AI: {req_err}. Falling back to simulation.")
            except Exception as e:
                print(f"An unexpected error occurred with Tripo AI API: {e}. Falling back to simulation.")

    # If API integration is not chosen, or it fails/times out, proceed with simulation
    if needs_simulation:
        print("Creating a simulated model as a fallback.")
        # For this simulation, we'll create a simple trimesh sphere or box as a stand-in.
        # The complexity of the model could be varied based on the prompt for a better demo.
        if "sun" in prompt.lower() or "portal" in prompt.lower() or "luminaries" in prompt.lower():
            print("Creating a simulated sphere for sun/portal/luminaries related prompt.")
            return trimesh.creation.icosphere(subdivisions=2, radius=1.0)
        elif "wind" in prompt.lower() or "storehouses" in prompt.lower():
            print("Creating a simulated box for wind/storehouses related prompt.")
            return trimesh.creation.box()
        else:
            print("Creating a generic simulated cylinder.")
            return trimesh.creation.cylinder(radius=1.0, height=2.0)
    return generated_model # Return the API generated model if successful, otherwise None if simulation was skipped but API also failed.

# Select one of the generated prompts for the proof of concept
# Let's pick the prompt describing the 'gates of the sun'.
proof_of_concept_prompt = ""
if len(prompts) > 2: # Check if there are at least 3 elements (index 0, 1, 2)
    proof_of_concept_prompt = prompts[2]
elif len(prompts) > 0: # If prompts exist but not enough for index 2, take the first one
    print("WARNING: Not enough prompts generated to select the third prompt (index 2). Selecting the first available prompt.")
    proof_of_concept_prompt = prompts[0]
else: # If prompts is completely empty
    print("ERROR: No prompts were generated. This likely means the text fetching or keyword extraction failed.")
    print("Falling back to a generic default prompt for simulation.")
    proof_of_concept_prompt = "3D environment of a simple geometric shape, such as a sphere."

print(f"\nSelected prompt for 3D generation:\n{proof_of_concept_prompt}\n")

# Simulate 3D model generation
generated_trimesh_model = generate_3d_model_from_api(proof_of_concept_prompt, api_choice='tripo')

# Print information about the generated model using available attributes
if generated_trimesh_model:
    print(f"\nSimulated 3D model generated: {type(generated_trimesh_model).__name__} with {len(generated_trimesh_model.vertices)} vertices and {len(generated_trimesh_model.faces)} faces.")
else:
    print("\nNo 3D model was generated from the API or simulation.")
```

## Creating a 3D Scatter Plot for Prompts

To visualize the `prompts` in a 3D scatter plot, we need numerical coordinates for each prompt. Since the prompts are text, we will derive some simple numerical features for demonstration purposes:

*   **X-axis**: Length of the prompt string.
*   **Y-axis**: Number of words in the prompt string.
*   **Z-axis**: A derived numerical value (e.g., sum of character ASCII values modulo a constant) to represent another dimension.

```python
import plotly.graph_objects as go

# The 'prompts' list is already generated and available from previous cells (e.g., eaf9d522).
# No need to redefine urls, get_text, enoch_text, extract_visual, sun_gates, or prompts here.

# Prepare data for the 3D scatter plot using the existing 'prompts' list
x_coords = [len(p) for p in prompts]
y_coords = [len(p.split()) for p in prompts]
z_coords = [sum(ord(char) for char in p) % 200 for p in prompts] # Simple derived Z-value

# Create the 3D scatter plot
fig = go.Figure(data=[
    go.Scatter3d(
        x=x_coords,
        y=y_coords,
        z=z_coords,
        mode='markers',
        marker=dict(
            size=8,
            color=z_coords, # Color points by their Z-value
            colorscale='Viridis', # Choose a colorscale
            opacity=0.8
        ),
        text=[f'Prompt: {p}<br>Length: {l}<br>Words: {w}<br>Z-Value: {z_val}' for p, l, w, z_val in zip(prompts, x_coords, y_coords, z_coords)],
        hoverinfo='text'
    )
])

# Update layout for better visualization
fig.update_layout(
    title='3D Scatter Plot of Generated Prompts',
    scene=dict(
        xaxis_title='Prompt Length (characters)',
        yaxis_title='Prompt Word Count',
        zaxis_title='Derived Z-Value',
        aspectmode='cube'
    ),
    margin=dict(l=0, r=0, b=0, t=40)
)

fig.show()

print("Interactive 3D scatter plot of prompts displayed above. Hover over points to see details.")
```
